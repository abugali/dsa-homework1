````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
cities_clean = []
for i in range(len(cities)):
    cities_clean.append(cities[i].capitalize().replace(" ", ""))
amounts_int = []
for j in amounts:
    amounts_int.append(int(j))
count_checks, count_sum, count_avg, count_paid = 0, 0, 0, 0
for i in amounts_int:
    count_checks +=1
    count_sum += i
count_avg = round(count_sum/count_checks)
for i in paid:
    if i == True:
        count_paid += 1
for city, amount, status in zip(cities_clean, amounts_int, paid):
    print(f"{city} | {amount} | {status}")
````

````python
# Стандарт
````

````python
revenue_almaty = 0
revenue_astana = 0
for city, amount, status in zip(cities_clean, amounts_int, paid):
    if status == True:
        if city == "Almaty":
            revenue_almaty += amount
        elif city == "Astana":
            revenue_astana += amount
print(revenue_almaty, revenue_astana)
max_paid_amount = 0
max_paid_index = -1
max_paid_city = ""
for i in range(len(amounts_int)):
    if paid[i] == True and amounts_int[i] > max_paid_amount:
        max_paid_amount = amounts_int[i]
        max_paid_index = i
        max_paid_city = cities_clean[i]
print(max_paid_index, max_paid_city)
weekend_count = 0
for day in days:
    if day == "сб" or day == "вс":
        weekend_count += 1
print((weekend_count / len(days)) * 100)
savings = 0
months = 0
course_price = 400000
monthly_deposit = 15000
while savings < course_price:
    savings += monthly_deposit
    months += 1
print(months)
savings_inc = 0
months_inc = 0
current_deposit = 15000
while savings_inc < course_price:
    savings_inc += current_deposit
    months_inc += 1
    current_deposit += 1000
print(months_inc)
````
